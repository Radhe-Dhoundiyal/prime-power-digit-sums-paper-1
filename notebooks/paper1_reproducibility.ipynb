## Exact normalization

The authoritative statistic is $\widetilde A_p(n)=S(p^n)/\ell_p(n)$, where $\ell_p(n)=\lfloor n\log_{10}p\rfloor+1$. The implementation uses `len(str(power))` as the exact digit count and independently checks the logarithmic formula. The older $A_p(n)=S(p^n)/(n\log_{10}p)$ normalization is supplementary only.
